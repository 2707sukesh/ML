```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
arr = np.linspace(0, 5, num=1000)
## here since thelse elements are equal spaced, their standard deviation should almost be close to zero.

print('mean , aka average', arr.mean())
print('standard deviation', arr.std())
arr
```

```python
normal_dist = np.random.normal(loc=50, scale=5, size=10000) # loc = 0, means the mean
normal_dist
mu = normal_dist.mean()
sigma = normal_dist.std()
print('mean, mu', normal_dist.mean())
print('standard deviation ', normal_dist.std())
x = np.linspace(mu-2*sigma, mu+2*sigma, 100)
bell_curve = stats.norm.pdf(x, mu, sigma) 
## we give these values to x, mu, sigma, and this function transforms x, such that 
# if x is closer to mu, then it returns higher probabilyt and away from mu returns lower probabiilty, 
# ideally this takes the values of x and generates the value of y.
plt.hist(normal_dist, bins=1000, density=True, alpha=0.6, color='skyblue', label='Sample Data Histogram')

plt.plot(x, bell_curve, color='darkblue', linewidth=2, label='Fitted Bell curve') # we can directly plot the histogram.
plt.title('Bell Graph (Normal Distrubution) fitted to sample data')
plt.xlabel('Values')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(axis='y', alpha = 0.3)
plt.show()
```
